## *cross-verification (anti-spoofing)*

Esegue le sfide di ogni chip arruolato sotto il noise model di ogni altro chip e costruisce la matrice di Fidelity. La diagonale (chip legittimo) deve avere Fidelity alta e superare la soglia; le celle fuori diagonale (tentativi di spoofing su hardware diverso) devono restare sotto la soglia ed essere rifiutate.
